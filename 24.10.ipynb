```python
# 1. Valideyn (Super) sinif
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def get_details(self):
        return f"Ad: {self.name}, Maaş: {self.salary} AZN"

    def work(self):
        return f"{self.name} ümumi işləri yerinə yetirir."


# 2. Uşaq (Sub) sinif - Manager (Miras alır və super() istifadə edir)
class Manager(Employee):
    def __init__(self, name, salary, department):
        # Valideyn sinfin __init__ metodunu çağırırıq (super() vasitəsilə)
        super().__init__(name, salary)
        self.department = department

    # Metodun ləğvi (Method Overriding) və super() istifadəsi
    def get_details(self):
        # Valideynin metodundan gələn məlumatı alırıq
        base_details = super().get_details()
        return f"{base_details}, Şöbə: {self.department}"

    # Valideynin work metodunu özünə uyğun dəyişirik (Overriding)
    def work(self):
        return f"Menecer {self.name} {self.department} şöbəsini idarə edir."


# 3. Başqa bir Uşaq sinif - Developer
class Developer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language

    def work(self):
        return f"Proqramçı {self.name} {self.language} dilində kod yazır."


# --- Polimorfizm Nümayişi ---
# Bu funksiya obyektin hansı sinifdən olduğunu bilmədən sadəcə work() metodunu çağırır
def show_employee_work(employee_obj):
    print(employee_obj.work())


# --- Obyektlərin Yaradılması və Test Edilməsi ---

# Obyektləri yaradırıq
mgr = Manager("Elmir", 3500, "IT")
dev = Developer("Aytən", 2800, "Python")

print("--- Məlumatlar ---")
print(mgr.get_details()) 
print(dev.get_details()) 

print("\n--- Polimorfizm Nümunəsi ---")
show_employee_work(mgr)  # Manager sinfinin work() metodu işə düşür
show_employee_work(dev)  # Developer sinfinin work() metodu işə düşür
```

```python
class Rectangle:
	def __init__(self, a, b):
		self.a = a
		self.b = b

	def compute_area1(a, b):
		return a * b
	
	@staticmethod
	def compute_area2(a, b):
		return a * b

r = Rectangle(3,4)

Rectangle.compute_area1(3,4) # 12
Rectangle.compute_area2(3,4) # 12
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_birth_year(cls, name, birth_year):
        age = 2026 - birth_year
        return cls(name, age)

person = Person.from_birth_year("ozee", 2008)
print(person.name, person.age) # Alice 30
```

```python
class User:
    total_users = 0

    def __init__(self, name):
        self.name = name
        User._increment_user_count()

    @classmethod
    def _increment_user_count(cls):
        cls.total_users += 1

u1 = User("Alice")
u2 = User("Bob")
User.total_users # 2
```

```python
class Rectangle:
	def __init__(self, a, b):
		self.a = a
		self.b = b
	
	@staticmethod
	def compute_area(a, b):
		return a * b

class Square(Rectangle):
	def __init__(self, a):
		super().__init__(a, a)

	@classmethod
	def compute_area(cls, a):
		return super(Square, cls).compute_area(a, a)

Square.compute_area(3) # 9
```

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value

    @radius.deleter
    def radius(self):
        del self._radius

c = Circle(5)
print(c.radius)      # Accessing like an attribute (getter)
c.radius = 7         # Setting like an attribute (setter)
print(c.radius)
del c.radius         # Deleting like an attribute (deleter)
```
